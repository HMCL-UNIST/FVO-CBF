# Runtime Parameters (ACS controller + QP solver)

These parameters do NOT affect initial condition generation — they are used only during the
simulation runtime. Edit values here and **Run All** to update `parameter/*.npy` without
re-generating initial conditions.


```python
import numpy as np
import os

os.makedirs('parameter', exist_ok=True)
```

```python
# --- ACS parameters ---
beta = 1/4
lamda = 1
k1 = 0.3
k2 = 0.02
```

```python
# --- QP solver parameters ---
qp_max_iter = 100
qp_eps_abs = 1e-6
qp_time_limit = 0.05   # seconds per QP solve; 0 = no limit
fi_threshold = -1e-4   # forward-invariance failure threshold
```

```python
# --- Save to parameter/*.npy ---
np.save('parameter/beta.npy', beta)
np.save('parameter/lamda.npy', lamda)
np.save('parameter/k1.npy', k1)
np.save('parameter/k2.npy', k2)
np.save('parameter/qp_max_iter.npy', qp_max_iter)
np.save('parameter/qp_eps_abs.npy', qp_eps_abs)
np.save('parameter/qp_time_limit.npy', qp_time_limit)
np.save('parameter/fi_threshold.npy', fi_threshold)

print('Saved runtime params:')
for k in ['beta','lamda','k1','k2','qp_max_iter','qp_eps_abs','qp_time_limit','fi_threshold']:
    v = float(np.load(f'parameter/{k}.npy'))
    print(f'  {k} = {v}')
```
